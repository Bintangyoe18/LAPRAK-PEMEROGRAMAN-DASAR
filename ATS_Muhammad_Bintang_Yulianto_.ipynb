```python
# Challenge: Pemantauan Beban Listrik Harian
#
# Masalah:
# Sebuah gedung mencatat beban listrik (kW) tiap jam. Kadang sensor
# mengirim data rusak ("ERR"). Program ini mengelompokkan beban tiap jam
# (rendah/normal/puncak), mengabaikan data rusak, lalu menghitung total
# energi, rata-rata beban, beban tertinggi, dan perkiraan biaya listrik.
#
# Cara kerja:
# 1. Data beban disimpan dalam list (bertipe teks, seperti dari sensor).
# 2. Perulangan for membaca data satu per satu.
# 3. try-except mengubah teks jadi angka; kalau gagal, data dilewati.
# 4. if-elif-else menentukan status beban.
# 5. Setelah loop, hasil dihitung dan ditampilkan.

data_beban = ["35.5", "42.0", "ERR", "58.3", "76.8", "91.2", "64.5", "48.9"]

batas_puncak = 80    # kW, di atas/sama dengan ini = beban puncak
batas_normal = 50    # kW, di atas/sama dengan ini = normal
tarif = 1444.70      # rupiah per kWh

beban_valid = []     # list untuk menyimpan data yang valid
tidak_valid = 0
jumlah_rendah = 0
jumlah_normal = 0
jumlah_puncak = 0
beban_maks = 0
jam_maks = 0

print("=== Pemantauan Beban Listrik Harian ===")

for jam, teks in enumerate(data_beban, start=1):
    # Ubah teks jadi angka, kalau gagal lewati data ini
    try:
        beban = float(teks)
    except ValueError:
        print(f"Jam {jam}: data tidak valid")
        tidak_valid += 1
        continue

    beban_valid.append(beban)

    # Tentukan status beban
    if beban >= batas_puncak:
        status = "BEBAN PUNCAK"
        jumlah_puncak += 1
    elif beban >= batas_normal:
        status = "NORMAL"
        jumlah_normal += 1
    else:
        status = "RENDAH"
        jumlah_rendah += 1

    # Catat beban tertinggi
    if beban > beban_maks:
        beban_maks = beban
        jam_maks = jam

    print(f"Jam {jam}: {beban} kW -> {status}")

# Hitung hasil akhir
jumlah_valid = len(beban_valid)

if jumlah_valid > 0:
    total_energi = sum(beban_valid)           # tiap data = 1 jam, jadi kW x 1 jam = kWh
    rata_rata = total_energi / jumlah_valid
    biaya = total_energi * tarif

    print("\n--- Ringkasan ---")
    print(f"Data valid          : {jumlah_valid}")
    print(f"Data tidak valid    : {tidak_valid}")
    print(f"Status rendah/normal/puncak: {jumlah_rendah}/{jumlah_normal}/{jumlah_puncak}")
    print(f"Total energi        : {total_energi:.2f} kWh")
    print(f"Rata-rata beban     : {rata_rata:.2f} kW")
    print(f"Beban tertinggi     : {beban_maks} kW (jam {jam_maks})")
    print(f"Perkiraan biaya     : Rp {biaya:,.0f}")
else:
    print("Tidak ada data valid yang bisa dihitung.")
```
